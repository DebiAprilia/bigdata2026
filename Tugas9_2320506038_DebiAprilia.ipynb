Platform "Kopi Nusantara" (skenario yang sama dari Tugas Mandiri Pertemuan 7) ingin mulai menyimpan **katalog menu** mereka di MongoDB — karena setiap cabang punya menu musiman/spesial dengan atribut yang berbeda-beda (tidak semua menu punya atribut yang sama, misalnya hanya menu kopi yang punya atribut "tingkat_kepahitan", sementara menu makanan punya atribut "pedas_level"). **A. Membangun Katalog (Create)** *(bobot 20%)*

Buat database baru bernama `kopi_nusantara`, collection bernama `menu`. Sisipkan **minimal 12 dokumen** menu dengan field bervariasi sesuai kategorinya:
- Menu kategori **"Kopi"**: wajib memiliki field `nama_menu`, `harga`, `kategori`, dan `tingkat_kepahitan` (skala 1-5).
- Menu kategori **"Makanan"**: wajib memiliki field `nama_menu`, `harga`, `kategori`, dan `pedas_level` (skala 0-3).
- Boleh menambahkan field lain sesuai kreativitas anda (mis. `stok`, `rekomendasi_barista`).

```python
from pymongo import MongoClient

#Membuat koneksi ke MongoDB
client = MongoClient("mongodb://localhost:27017/")

#Membuat database baru bernama 'kopi_nusantara'
db = client["kopi_nusantara"]

#Membuat collection bernama 'menu'
koleksi_menu = db["menu"]

#Menyisipkan 12 dokumen menu menggunakan 'insert_many'
koleksi_menu.insert_many([
    {"nama_menu": "Americano", "harga": 17000, "kategori": "Kopi", "tingkat_kepahitan": 5},
    {"nama_menu": "Latte", "harga": 21000, "kategori": "Kopi", "tingkat_kepahitan": 2},
    {"nama_menu": "Matcha Espresso", "harga": 25000, "kategori": "Kopi", "tingkat_kepahitan": 4},
    {"nama_menu": "Butterscotch Aren Latte", "harga": 19000, "kategori": "Kopi", "tingkat_kepahitan": 2},
    {"nama_menu": "Cappucino", "harga": 22000, "kategori": "Kopi", "tingkat_kepahitan": 3},
    {"nama_menu": "Caramel Macchiato", "harga": 28000, "kategori": "Kopi", "tingkat_kepahitan": 1},
    {"nama_menu": "Choco Chip Cookies", "harga": 14000, "kategori": "Makanan", "pedas_level": 0},
    {"nama_menu": "Vanilla Choux Puff", "harga": 16000, "kategori": "Makanan", "pedas_level": 0},
    {"nama_menu": "Chocolate Donut", "harga": 13000, "kategori": "Makanan", "pedas_level": 0},
    {"nama_menu": "Chicken Wings", "harga": 35000, "kategori": "Makanan", "pedas_level": 3},
    {"nama_menu": "Pasta", "harga": 45000, "kategori": "Makanan", "pedas_level": 2},
    {"nama_menu": "French Fries", "harga": 23000, "kategori": "Makanan", "pedas_level": 2}
])

print("Koneksi ke MongoDB berhasil!")
print("Daftar database yang ada:", client.list_database_names())
print("Berhasil menambahkan 12 Menu!")
```

**B. Kueri Bertingkat (Read)** *(bobot 20%)*

Tampilkan hasil dari **tiga** kueri berikut :
1. Seluruh menu kategori "Kopi" dengan `tingkat_kepahitan` di atas 3.
2. Seluruh menu dengan `harga` di antara 15000 dan 30000.
3. Menu yang **tidak memiliki** field `pedas_level` sama sekali.

```python
#Menampilkan seluruh menu kategori 'Kopi' dengan 'tingkat_kepahitan' diatas 3
print("1. Kopi dengan tingkat kepahitan diatas 3:")
for m in koleksi_menu.find({"kategori": "Kopi", "tingkat_kepahitan": {"$gt": 3}}):
    print(f" - {m['nama_menu']}")

#Menampilkan seluruh menu dengan harga di antara 15000 dan 30000
print("\n2. Menu dengan harga di antara 15000 dan 30000:")
for m in koleksi_menu.find({"harga": {"$gte": 15000, "$lte": 30000}}):
    print(f" - {m['nama_menu']} (Rp{m['harga']})")

#Menampilkan menu yang tidak memiliki field 'pedas_level'
print("\n3. Menu yang tidak memiliki field pedas_level:")
for m in koleksi_menu.find({"pedas_level": {"$exists": False}}):
    print(f" - {m['nama_menu']}")
```

**C. Pembaruan Data (Update)** *(bobot 15%)*

Tambahkan field `status` bernilai `"Tersedia"` ke seluruh dokumen menggunakan `update_many()`. Kemudian, ubah `status` menjadi `"Habis"` khusus untuk 2 menu pilihan anda menggunakan `update_one()`.

```python
#Menambahkan field status bernilai "Tersedia" ke seluruh dokumen
koleksi_menu.update_many({}, {"$set": {"status": "Tersedia"}})

#Mengubah status menjadi Habis untuk 2 menu pilihan
koleksi_menu.update_one({"nama_menu": "Latte"}, {"$set": {"status": "Habis"}})
koleksi_menu.update_one({"nama_menu": "Pasta"}, {"$set": {"status": "Habis"}})

print("Pembaruan data selesai.")
```

**D. Integrasi ke Spark** *(bobot 25%)*

Ambil seluruh data dari collection `menu`, ubah menjadi Spark DataFrame (ikuti pola Sub-bab 9.6 — perhatikan field yang mungkin kosong/`null` pada sebagian dokumen, seperti `tingkat_kepahitan` yang hanya dimiliki menu Kopi). Tampilkan:
1. `printSchema()` hasil konversi.
2. Rata-rata `harga` per `kategori` menggunakan `groupBy`.

```python
from pyspark.sql import SparkSession
import pandas as pd
from pyspark.sql.functions import avg

#Inisialisasi Spark
spark = SparkSession.builder \
    .appName("Tugas9-MongoDBSpark") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

#Mengambil seluruh data dari collection Menu dan ubah ke pandas
dokumen_menu = list(koleksi_menu.find({}))
df_pandas = pd.DataFrame(dokumen_menu)

#Mengubah tipedata ObjectId menjadi string agar terbaca oleh spark
df_pandas["_id"] = df_pandas["_id"].astype(str)

#Konversi ke Spark DataFrame
df_spark_menu = spark.createDataFrame(df_pandas)

#Menampilkan schema hasil konversi
print("1. Schema DataFrame:")
df_spark_menu.printSchema()

#Menghitung rata rata harga per kategori
print("\n2. Rata - rata harga per kategori:")
df_spark_menu.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga")
).show()
```

```python
#Menutup sparksession
spark.stop()
print("SparkSession ditutup.")
```

Tulis pada markdown cell (**minimal 100 kata**): apa tantangan yang anda temui saat mengonversi data MongoDB (dengan field yang tidak seragam antar dokumen) menjadi Spark DataFrame (yang mengharuskan skema tabel seragam)? Bagaimana anda mengatasinya?

**Jawaban:**
Tantangan yang ditemui saat mengonversi data MongoDB ke Spark DataFrame adalah pada perbedaan skema keduanya, jika pada MongoDB skemanya bersifat fleksibel sehingga ketika diubah ke Spark otomatis akan membuat semua kolom untuk semua field yang ada (misalnya pada kolom 'tingkat_kepahitan' dan 'pedas_level' akan dibuat semuanya), sedangkan Spark DataFrame skemanya tabelnya seragam. Tantangan yang lain adalah dokumen kategori Makanan akan menghasilkan nilai null di kolom 'tingkat_kepahitan', serta kategori Minuman akan menghasilkan nilai null di kolom 'pedas_level, bahkan hal tersebut juga dapat menyebabkan error runtime karena tipe datanya tidak cocok. Untuk mengatasi tantangan tersebut, bisa menggunakan Pandas untuk menstandarkan tipe data atau bisa menggunakan fungsi fillna() di PySPark untuk mengisi nilai null agar analisis agregasi tidak terganggu oleh nilai tersebut.
